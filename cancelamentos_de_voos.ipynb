==============================
MAPEAMENTO DE COLUNAS – ANAC & BTS
Projeto: Previsão de Atrasos de Voos
==============================

OBJETIVO:
Padronizar os dados ANAC (Brasil) e BTS (EUA) para uso conjunto em um único modelo,
utilizando apenas informações disponíveis ANTES da decolagem (evitar data leakage).

---------------------------------
COLUNAS PADRÃO DO MODELO
---------------------------------

companhia
origem
destino
data_voo
hora_prevista
dia_semana
distancia_km
atrasado  (TARGET)

---------------------------------
MAPEAMENTO – ANAC (Brasil)
---------------------------------

ICAO Empresa Aérea        -> companhia
ICAO Aeródromo Origem    -> origem
ICAO Aeródromo Destino   -> destino
Partida Prevista         -> data_voo + hora_prevista
Partida Prevista         -> dia_semana (derivado)
Partida Real             -> usado apenas para criar "atrasado"
Situação Voo             -> apoio / validação

Regra atraso ANAC:
atrasado = Partida Real > Partida Prevista

---------------------------------
MAPEAMENTO – BTS (EUA)
---------------------------------

Reporting_Airline        -> companhia
Origin                   -> origem
Dest                     -> destino
FlightDate               -> data_voo
CRSDepTime               -> hora_prevista
FlightDate               -> dia_semana (derivado)
Distance                 -> distancia_km
DepDelay                 -> usado apenas para criar "atrasado"

Regra atraso BTS:
atrasado = DepDelay > 0

---------------------------------
COLUNAS DESCARTADAS – ANAC

Código Tipo Linha

- Informações administrativas
- Pós-voo

OBSERVAÇÃO IMPORTANTE
---------------------------------

O dataset BTS possui mais de 100 colunas, mas apenas uma fração
