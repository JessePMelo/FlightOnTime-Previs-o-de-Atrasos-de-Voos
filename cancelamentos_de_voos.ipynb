#####ANAC

```python
import pandas as pd

pd.set_option("display.max_columns", None)

BASE_URL = "https://github.com/JessePMelo/FlightOnTime-Previs-o-de-Atrasos-de-Voos/releases/download/Data_FlightOnTime_v1/"
arquivo = "anac_vra_2021_01.csv"
url = BASE_URL + arquivo

print(f"Lendo {arquivo}")

# ============================================================
# 1) Carregar dados
# ============================================================
df = pd.read_csv(
    url,
    sep=";",
    encoding="utf-8",
    header=1,
    low_memory=False
)

# ============================================================
# 2) Converter datas (garantir datetime)
# ============================================================
df["Partida Prevista"] = pd.to_datetime(df["Partida Prevista"], errors="coerce")
df["Partida Real"] = pd.to_datetime(df["Partida Real"], errors="coerce")

# remover linhas sem datas essenciais
df = df.dropna(subset=["Partida Prevista", "Partida Real"]).copy()

# ============================================================
# 3) Criar FEATURES (somente pré-voo)
# ============================================================

# --------------------------
# hora prevista (0..23)
# --------------------------
df["scheduled_hour"] = df["Partida Prevista"].dt.hour

# --------------------------
# dia da semana (0=Mon .. 6=Sun)
# --------------------------
df["day_of_week"] = df["Partida Prevista"].dt.dayofweek

# --------------------------
# mês do ano (1..12)
# --------------------------
df["month"] = df["Partida Prevista"].dt.month

# --------------------------
# posição da semana no mês (1..5)
# --------------------------
df["week_of_month"] = ((df["Partida Prevista"].dt.day - 1) // 7 + 1)

# --------------------------
# fim de semana
# --------------------------
df["is_weekend"] = df["day_of_week"].isin([5, 6]).astype(int)

# --------------------------
# faixa horária (bucket)
# --------------------------
def hour_bucket(h: int) -> int:
    if h < 6:
        return 0  # madrugada
    elif h < 12:
        return 1  # manhã
    elif h < 18:
        return 2  # tarde
    else:
        return 3  # noite

df["hour_bucket"] = df["scheduled_hour"].apply(hour_bucket)

# --------------------------
# rota (origem_destino)
# --------------------------
df["route"] = (
    df["ICAO Aeródromo Origem"].astype(str)
    + "_"
    + df["ICAO Aeródromo Destino"].astype(str)
)

# --------------------------
# horário de pico (geral)
# --------------------------
df["is_peak_hour"] = df["scheduled_hour"].isin(
    [6, 7, 8, 9, 17, 18, 19, 20]
).astype(int)

# --------------------------
# pico da manhã
# --------------------------
df["is_morning_peak"] = df["scheduled_hour"].isin([6, 7, 8, 9]).astype(int)

# --------------------------
# pico da tarde/noite
# --------------------------
df["is_evening_peak"] = df["scheduled_hour"].isin([17, 18, 19, 20]).astype(int)

# ============================================================
# 4) Blindagem final de features
# ============================================================
df = df.dropna(
    subset=[
        "scheduled_hour",
        "day_of_week",
        "month",
        "week_of_month",
        "is_weekend",
        "hour_bucket",
        "route",
        "is_peak_hour",
        "is_morning_peak",
        "is_evening_peak",
    ]
).copy()

# garantir tipos
cols_int = [
    "scheduled_hour",
    "day_of_week",
    "month",
    "week_of_month",
    "is_weekend",
    "hour_bucket",
    "is_peak_hour",
    "is_morning_peak",
    "is_evening_peak",
]
for c in cols_int:
    df[c] = df[c].astype(int)

# ============================================================
# 5) TARGET: atraso (PADRÃO INTERNACIONAL: >= 15 minutos)
# ============================================================
# 1) calcula atraso em minutos (pode ser negativo se saiu antes)
df["delay_minutes"] = (df["Partida Real"] - df["Partida Prevista"]).dt.total_seconds() / 60

# 2) blindagem: se por algum motivo virou NaN/inf, remove
df["delay_minutes"] = pd.to_numeric(df["delay_minutes"], errors="coerce")
df = df.dropna(subset=["delay_minutes"]).copy()

# 3) delayed = 1 se atraso >= 15 min, senão 0
df["delayed"] = (df["delay_minutes"] >= 15).astype(int)

# (opcional) se você não vai usar delay_minutes como feature, remove para não vazar info
df = df.drop(columns=["delay_minutes"])

# ============================================================
# 6) Renomear colunas (padrão do modelo)
# ============================================================
df = df.rename(columns={
    "ICAO Empresa Aérea": "airline",
    "ICAO Aeródromo Origem": "origin",
    "ICAO Aeródromo Destino": "destination"
})

# ============================================================
# 7) Normalização de texto (blindagem)
# ============================================================
for col in ["airline", "origin", "destination", "route"]:
    df[col] = df[col].astype(str).str.strip().str.upper()

# ============================================================
# 8) Dataset FINAL
# ============================================================
df_anac_2021_01 = df[
    [
        "airline",
        "origin",
        "destination",
        "route",
        "scheduled_hour",
        "hour_bucket",
        "is_peak_hour",
        "is_morning_peak",
        "is_evening_peak",
        "day_of_week",
        "is_weekend",
        "week_of_month",
        "month",
        "delayed",
    ]
].copy()

# ============================================================
# 9) Checagens finais
# ============================================================
print("\nNaNs por coluna (deve ser tudo 0):")
print(df_anac_2021_01.isna().sum())

print("\nDistribuição do target (delayed):")
print(df_anac_2021_01["delayed"].value_counts())

df_anac_2021_01.head()
```

```python
import pandas as pd

# visualizar todas as colunas (opcional)
pd.set_option("display.max_columns", None)

BASE_URL = "https://github.com/JessePMelo/FlightOnTime-Previs-o-de-Atrasos-de-Voos/releases/download/Data_FlightOnTime_v1/"
arquivo = "bts_ontime_2021_01.csv"
url = BASE_URL + arquivo

print(f"Lendo {arquivo}")

# ============================================================
# 1) Carregar dados
# ============================================================
df = pd.read_csv(
    url,
    sep=",",
    encoding="utf-8",
    low_memory=False
)

# ============================================================
# 2) REMOVER VOOS CANCELADOS (Cancelado ≠ atraso)
# ============================================================
df["Cancelled"] = pd.to_numeric(df["Cancelled"], errors="coerce")
df = df.dropna(subset=["Cancelled"]).copy()
df = df[df["Cancelled"] == 0].copy()

# ============================================================
# 3) Criar FEATURES (somente pré-voo)
# ============================================================

# --------------------------
# scheduled_hour (CRSDepTime vem como HHMM)
# --------------------------
df["CRSDepTime"] = pd.to_numeric(df["CRSDepTime"], errors="coerce")
df = df.dropna(subset=["CRSDepTime"]).copy()

df["scheduled_hour"] = (df["CRSDepTime"] // 100)
df = df[df["scheduled_hour"].between(0, 23)].copy()
df["scheduled_hour"] = df["scheduled_hour"].astype(int)

# --------------------------
# day_of_week (BTS: 1=Mon .. 7=Sun -> 0..6)
# --------------------------
df["DayOfWeek"] = pd.to_numeric(df["DayOfWeek"], errors="coerce")
df = df.dropna(subset=["DayOfWeek"]).copy()

df["day_of_week"] = df["DayOfWeek"] - 1
df = df[df["day_of_week"].between(0, 6)].copy()
df["day_of_week"] = df["day_of_week"].astype(int)

# --------------------------
# FlightDate -> datetime (para month/week_of_month)
# --------------------------
df["FlightDate"] = pd.to_datetime(df["FlightDate"], errors="coerce")
df = df.dropna(subset=["FlightDate"]).copy()

# month (1..12)
df["month"] = df["FlightDate"].dt.month.astype(int)

# week_of_month (1..5) (igual ANAC)
df["week_of_month"] = ((df["FlightDate"].dt.day - 1) // 7 + 1).astype(int)

# is_weekend
df["is_weekend"] = df["day_of_week"].isin([5, 6]).astype(int)

# hour_bucket
def hour_bucket(h: int) -> int:
    if h < 6:
        return 0  # madrugada
    elif h < 12:
        return 1  # manhã
    elif h < 18:
        return 2  # tarde
    else:
        return 3  # noite

df["hour_bucket"] = df["scheduled_hour"].apply(hour_bucket).astype(int)

# route (origem_destino)
df["route"] = df["Origin"].astype(str) + "_" + df["Dest"].astype(str)

# picos
df["is_peak_hour"] = df["scheduled_hour"].isin([6, 7, 8, 9, 17, 18, 19, 20]).astype(int)
df["is_morning_peak"] = df["scheduled_hour"].isin([6, 7, 8, 9]).astype(int)
df["is_evening_peak"] = df["scheduled_hour"].isin([17, 18, 19, 20]).astype(int)

# ============================================================
# 4) Blindagem final de features + campos essenciais de texto
# ============================================================
df = df.dropna(
    subset=[
        "Reporting_Airline", "Origin", "Dest",  # garante chaves essenciais
        "scheduled_hour",
        "day_of_week",
        "month",
        "week_of_month",
        "is_weekend",
        "hour_bucket",
        "route",
        "is_peak_hour",
        "is_morning_peak",
        "is_evening_peak",
    ]
).copy()

# garantir tipos
cols_int = [
    "scheduled_hour",
    "day_of_week",
    "month",
    "week_of_month",
    "is_weekend",
    "hour_bucket",
    "is_peak_hour",
    "is_morning_peak",
    "is_evening_peak",
]
for c in cols_int:
    df[c] = df[c].astype(int)

# ============================================================
# 5) TARGET: ATRASO (>= 15 min)  ✅ PADRÃO INTERNACIONAL
# ============================================================
df["DepDel15"] = pd.to_numeric(df["DepDel15"], errors="coerce")
df = df.dropna(subset=["DepDel15"]).copy()

# blindagem: manter só 0/1
df = df[df["DepDel15"].isin([0, 1])].copy()
df["delayed"] = df["DepDel15"].astype(int)

# ============================================================
# 6) Renomear colunas (padrão do modelo)
# ============================================================
df = df.rename(columns={
    "Reporting_Airline": "airline",
    "Origin": "origin",
    "Dest": "destination"
})

# ============================================================
# 7) Normalização de texto (igual ANAC)
# ============================================================
for col in ["airline", "origin", "destination", "route"]:
    df[col] = df[col].astype(str).str.strip().str.upper()

# ============================================================
# 8) Dataset FINAL (MESMO PADRÃO DO ANAC)
# ============================================================
df_bts_2021_01 = df[
    [
        "airline",
        "origin",
        "destination",
        "route",
        "scheduled_hour",
        "hour_bucket",
        "is_peak_hour",
        "is_morning_peak",
        "is_evening_peak",
        "day_of_week",
        "is_weekend",
        "week_of_month",
        "month",
        "delayed",
    ]
].copy()

# ============================================================
# 9) Checagens finais
# ============================================================
print("\nNaNs por coluna (deve ser tudo 0):")
print(df_bts_2021_01.isna().sum())

print("\nDistribuição do target (delayed):")
print(df_bts_2021_01["delayed"].value_counts())

df_bts_2021_01.head()
```

```python
import pandas as pd
from tqdm.auto import tqdm

pd.set_option("display.max_columns", None)

# ============================================================
# CONFIG (mesmo padrão BTS)
# ============================================================
BASE_URL = "https://github.com/JessePMelo/FlightOnTime-Previs-o-de-Atrasos-de-Voos/releases/download/Data_FlightOnTime_v1/"

anos = [2021, 2022, 2023, 2024, 2025]
meses = range(1, 13)

# Saída incremental (igual BTS)
OUT_CSV = "df_anac_full_2021_01_2025_09.csv"

# Colunas finais (mesmo padrão que você usa no BTS)
COLS_FINAL = [
    "airline",
    "origin",
    "destination",
    "route",
    "scheduled_hour",
    "hour_bucket",
    "is_peak_hour",
    "is_morning_peak",
    "is_evening_peak",
    "day_of_week",
    "is_weekend",
    "week_of_month",
    "month",
    "delayed",
]

# ============================================================
# FUNÇÃO AUXILIAR (mesma do seu script)
# ============================================================
def hour_bucket(h: int) -> int:
    if h < 6:
        return 0  # madrugada
    elif h < 12:
        return 1  # manhã
    elif h < 18:
        return 2  # tarde
    else:
        return 3  # noite


def processar_anac_mes(url: str) -> pd.DataFrame:
    """
    ANAC 1 mês -> devolve SOMENTE COLS_FINAL
    (cópia fiel do seu script que funciona no 2021_01)
    """

    # ============================================================
    # 1) Carregar dados (ANAC exige header=1)
    # ============================================================
    df = pd.read_csv(
        url,
        sep=";",
        encoding="utf-8",
        header=1,
        low_memory=False
    )

    # ============================================================
    # 2) Converter datas (garantir datetime)
    # ============================================================
    df["Partida Prevista"] = pd.to_datetime(df["Partida Prevista"], errors="coerce")
    df["Partida Real"] = pd.to_datetime(df["Partida Real"], errors="coerce")

    # remover linhas sem datas essenciais
    df = df.dropna(subset=["Partida Prevista", "Partida Real"]).copy()

    # ============================================================
    # 3) Criar FEATURES (somente pré-voo)  (IGUAL AO SEU)
    # ============================================================

    df["scheduled_hour"] = df["Partida Prevista"].dt.hour
    df["day_of_week"] = df["Partida Prevista"].dt.dayofweek
    df["month"] = df["Partida Prevista"].dt.month
    df["week_of_month"] = ((df["Partida Prevista"].dt.day - 1) // 7 + 1)

    df["is_weekend"] = df["day_of_week"].isin([5, 6]).astype(int)

    df["hour_bucket"] = df["scheduled_hour"].apply(hour_bucket)

    df["route"] = (
        df["ICAO Aeródromo Origem"].astype(str)
        + "_"
        + df["ICAO Aeródromo Destino"].astype(str)
    )

    df["is_peak_hour"] = df["scheduled_hour"].isin([6, 7, 8, 9, 17, 18, 19, 20]).astype(int)
    df["is_morning_peak"] = df["scheduled_hour"].isin([6, 7, 8, 9]).astype(int)
    df["is_evening_peak"] = df["scheduled_hour"].isin([17, 18, 19, 20]).astype(int)

    # ============================================================
    # 4) Blindagem final de features (IGUAL AO SEU)
    # ============================================================
    df = df.dropna(
        subset=[
            "scheduled_hour",
            "day_of_week",
            "month",
            "week_of_month",
            "is_weekend",
            "hour_bucket",
            "route",
            "is_peak_hour",
            "is_morning_peak",
            "is_evening_peak",
        ]
    ).copy()

    cols_int = [
        "scheduled_hour",
        "day_of_week",
        "month",
        "week_of_month",
        "is_weekend",
        "hour_bucket",
        "is_peak_hour",
        "is_morning_peak",
        "is_evening_peak",
    ]
    for c in cols_int:
        df[c] = df[c].astype(int)

    # ============================================================
    # 5) TARGET: atraso (>= 15 minutos)  (IGUAL AO SEU)
    # ============================================================
    df["delay_minutes"] = (
        (df["Partida Real"] - df["Partida Prevista"])
        .dt.total_seconds() / 60
    )

    df["delay_minutes"] = pd.to_numeric(df["delay_minutes"], errors="coerce")
    df = df.dropna(subset=["delay_minutes"]).copy()

    df["delayed"] = (df["delay_minutes"] >= 15).astype(int)

    # remove delay_minutes (igual seu script)
    df = df.drop(columns=["delay_minutes"])

    # ============================================================
    # 6) Renomear colunas (padrão BTS)
    # ============================================================
    df = df.rename(columns={
        "ICAO Empresa Aérea": "airline",
        "ICAO Aeródromo Origem": "origin",
        "ICAO Aeródromo Destino": "destination"
    })

    # ============================================================
    # 7) Normalização de texto (igual seu script)
    # ============================================================
    for col in ["airline", "origin", "destination", "route"]:
        df[col] = df[col].astype(str).str.strip().str.upper()

    # ============================================================
    # 8) Retornar somente colunas finais
    # ============================================================
    return df[COLS_FINAL].copy()


# ============================================================
# GERAR LISTA DE ARQUIVOS (2021_01 → 2025_09)
# ============================================================
arquivos = []
for ano in anos:
    for mes in meses:
        if ano == 2025 and mes > 9:
            break
        arquivos.append(f"anac_vra_{ano}_{mes:02d}.csv")

print(f"Total de arquivos ANAC para processar: {len(arquivos)}")
print(f"Saída incremental: {OUT_CSV}")

# ============================================================
# LOOP COM ESCRITA INCREMENTAL (padrão BTS)
# ============================================================
erros = []
primeiro = True  # controla header no CSV final

for nome_arquivo in tqdm(arquivos, desc="Processando ANAC (2021_01 → 2025_09)"):
    url = BASE_URL + nome_arquivo

    try:
        df_mes = processar_anac_mes(url)

        df_mes.to_csv(
            OUT_CSV,
            mode="w" if primeiro else "a",
            header=primeiro,
            index=False
        )
        primeiro = False

        del df_mes  # solta RAM do mês

    except Exception as e:
        erros.append((nome_arquivo, str(e)))

# ============================================================
# RELATÓRIO FINAL (igual BTS)
# ============================================================
print("\n✅ ANAC final salvo em:", OUT_CSV)

if len(erros) > 0:
    print(f"\n⚠️ Arquivos com erro: {len(erros)}")
    print("Exemplos (até 5):")
    for item in erros[:5]:
        print(item)
else:
    print("\n✅ Nenhum erro nos arquivos.")

print("\nPrévia do arquivo final (5 linhas):")
print(pd.read_csv(OUT_CSV, nrows=5))
```

```python
import pandas as pd
from tqdm.auto import tqdm

pd.set_option("display.max_columns", None)

# ============================================================
# CONFIG
# ============================================================
BASE_URL = "https://github.com/JessePMelo/FlightOnTime-Previs-o-de-Atrasos-de-Voos/releases/download/Data_FlightOnTime_v1/"

anos = [2021, 2022, 2023, 2024, 2025]
meses = range(1, 13)

# Saída final (um CSV único, montado aos poucos)
OUT_CSV = "df_bts_full_2021_01_2025_09.csv"

# Colunas finais (padrão ANAC)
COLS_FINAL = [
    "airline",
    "origin",
    "destination",
    "route",
    "scheduled_hour",
    "hour_bucket",
    "is_peak_hour",
    "is_morning_peak",
    "is_evening_peak",
    "day_of_week",
    "is_weekend",
    "week_of_month",
    "month",
    "delayed",
]

# Ler só colunas necessárias (economiza RAM/tempo)
USECOLS = [
    "Reporting_Airline",
    "Origin",
    "Dest",
    "Cancelled",
    "CRSDepTime",
    "DayOfWeek",
    "FlightDate",
    "DepDel15",
]

# ============================================================
# FUNÇÃO AUXILIAR
# ============================================================
def hour_bucket(h: int) -> int:
    if h < 6:
        return 0
    elif h < 12:
        return 1
    elif h < 18:
        return 2
    else:
        return 3


def processar_bts_mes(url: str) -> pd.DataFrame:
    """
    Processa 1 CSV mensal do BTS e devolve apenas COLS_FINAL (padrão ANAC).
    Mantém somente variáveis pré-voo + target delayed (>=15 min).
    """

    # 1) Ler só colunas necessárias
    df = pd.read_csv(
        url,
        sep=",",
        encoding="utf-8",
        low_memory=False,
        usecols=USECOLS
    )

    # 2) Remover cancelados (cancelado ≠ atraso)
    df["Cancelled"] = pd.to_numeric(df["Cancelled"], errors="coerce")
    df = df.dropna(subset=["Cancelled"])
    df = df[df["Cancelled"] == 0].copy()

    # 3) scheduled_hour (CRSDepTime = HHMM)
    df["CRSDepTime"] = pd.to_numeric(df["CRSDepTime"], errors="coerce")
    df = df.dropna(subset=["CRSDepTime"]).copy()

    df["scheduled_hour"] = (df["CRSDepTime"] // 100)
    df = df[df["scheduled_hour"].between(0, 23)].copy()
    df["scheduled_hour"] = df["scheduled_hour"].astype(int)

    # 4) day_of_week (1..7 -> 0..6)
    df["DayOfWeek"] = pd.to_numeric(df["DayOfWeek"], errors="coerce")
    df = df.dropna(subset=["DayOfWeek"]).copy()

    df["day_of_week"] = df["DayOfWeek"] - 1
    df = df[df["day_of_week"].between(0, 6)].copy()
    df["day_of_week"] = df["day_of_week"].astype(int)

    # 5) FlightDate -> month + week_of_month
    df["FlightDate"] = pd.to_datetime(df["FlightDate"], errors="coerce")
    df = df.dropna(subset=["FlightDate"]).copy()

    df["month"] = df["FlightDate"].dt.month.astype(int)
    df["week_of_month"] = ((df["FlightDate"].dt.day - 1) // 7 + 1).astype(int)

    # 6) is_weekend
    df["is_weekend"] = df["day_of_week"].isin([5, 6]).astype(int)

    # 7) hour_bucket
    df["hour_bucket"] = df["scheduled_hour"].apply(hour_bucket).astype(int)

    # 8) route + picos
    df["route"] = df["Origin"].astype(str) + "_" + df["Dest"].astype(str)
    df["is_peak_hour"] = df["scheduled_hour"].isin([6, 7, 8, 9, 17, 18, 19, 20]).astype(int)
    df["is_morning_peak"] = df["scheduled_hour"].isin([6, 7, 8, 9]).astype(int)
    df["is_evening_peak"] = df["scheduled_hour"].isin([17, 18, 19, 20]).astype(int)

    # 9) TARGET delayed (>=15 min) usando DepDel15 (padrão internacional)
    df["DepDel15"] = pd.to_numeric(df["DepDel15"], errors="coerce")
    df = df.dropna(subset=["DepDel15"]).copy()
    df = df[df["DepDel15"].isin([0, 1])].copy()
    df["delayed"] = df["DepDel15"].astype(int)

    # 10) Renomear (padrão ANAC)
    df = df.rename(columns={
        "Reporting_Airline": "airline",
        "Origin": "origin",
        "Dest": "destination",
    })

    # 11) Normalização de texto (igual ANAC)
    for col in ["airline", "origin", "destination", "route"]:
        df[col] = df[col].astype(str).str.strip().str.upper()

    # 12) Retornar somente colunas finais
    return df[COLS_FINAL].copy()


# ============================================================
# GERAR LISTA DE ARQUIVOS
# ============================================================
arquivos = []
for ano in anos:
    for mes in meses:
        if ano == 2025 and mes > 9:
            break
        arquivos.append(f"bts_ontime_{ano}_{mes:02d}.csv")

print(f"Total de arquivos BTS para processar: {len(arquivos)}")
print(f"Saída incremental: {OUT_CSV}")

# ============================================================
# LOOP COM ESCRITA INCREMENTAL (SEM ESTOURAR RAM)
# ============================================================
erros = []
primeiro = True  # controla header no CSV final

for nome_arquivo in tqdm(arquivos, desc="Processando BTS (2021_01 → 2025_09)"):
    url = BASE_URL + nome_arquivo

    try:
        df_mes = processar_bts_mes(url)

        # grava no CSV final:
        # - primeiro mês: cria arquivo com header
        # - demais meses: append sem header
        df_mes.to_csv(
            OUT_CSV,
            mode="w" if primeiro else "a",
            header=primeiro,
            index=False
        )
        primeiro = False

        # liberar memória do mês atual
        del df_mes

    except Exception as e:
        erros.append((nome_arquivo, str(e)))

# ============================================================
# RELATÓRIO FINAL
# ============================================================
print("\n✅ BTS final salvo em:", OUT_CSV)

if len(erros) > 0:
    print(f"\n⚠️ Arquivos com erro: {len(erros)}")
    print("Exemplos (até 5):")
    for item in erros[:5]:
        print(item)
else:
    print("\n✅ Nenhum erro nos arquivos.")

# (opcional) ler uma amostra pequena só pra conferir sem estourar RAM
print("\nPrévia do arquivo final (5 linhas):")
print(pd.read_csv(OUT_CSV, nrows=5))
```

####Unificar as bases de dados !

Mesmo tentando otimizar os dados o volume de dados é muito grande mesmo optando por modificar de csv para parquet fiquei com receio de estourar ram. Me foi recomendado usar o chunks tambem !

```python
# ============================================================
# UNIFICAÇÃO FINAL: ANAC + BTS → PARQUET (SEM ESTOURAR RAM)
# - Lê os 2 CSVs finais em CHUNKS
# - Mantém exatamente o MESMO padrão de colunas
# - Barra de progresso geral + barra por dataset
# - Salva diretamente em Parquet
# ============================================================

import pandas as pd
from tqdm.auto import tqdm
import pyarrow as pa
import pyarrow.parquet as pq
import os

pd.set_option("display.max_columns", None)

# ============================================================
# CONFIG
# ============================================================
ANAC_CSV = "df_anac_full_2021_01_2025_09.csv"
BTS_CSV  = "df_bts_full_2021_01_2025_09.csv"

OUT_PARQUET = "df_full_anac_bts_2021_01_2025_09.parquet"

CHUNK_SIZE = 300_000  # seguro para Colab

COLS_FINAL = [
    "airline",
    "origin",
    "destination",
    "route",
    "scheduled_hour",
    "hour_bucket",
    "is_peak_hour",
    "is_morning_peak",
    "is_evening_peak",
    "day_of_week",
    "is_weekend",
    "week_of_month",
    "month",
    "delayed",
]

# ============================================================
# FUNÇÃO: CSV → PARQUET (INCREMENTAL)
# ============================================================
def csv_to_parquet_incremental(csv_path, parquet_writer=None):

    # contar chunks (barra correta)
    total_chunks = sum(
        1 for _ in pd.read_csv(csv_path, chunksize=CHUNK_SIZE)
    )

    reader = pd.read_csv(
        csv_path,
        usecols=COLS_FINAL,
        chunksize=CHUNK_SIZE,
        low_memory=False
    )

    for chunk in tqdm(
        reader,
        total=total_chunks,
        desc=f"Processando {os.path.basename(csv_path)}",
        leave=False
    ):
        table = pa.Table.from_pandas(chunk, preserve_index=False)

        if parquet_writer is None:
            parquet_writer = pq.ParquetWriter(
                OUT_PARQUET,
                table.schema,
                compression="snappy"
            )

        parquet_writer.write_table(table)

        # libera RAM imediatamente
        del chunk, table

    return parquet_writer


# ============================================================
# EXECUÇÃO
# ============================================================
if os.path.exists(OUT_PARQUET):
    os.remove(OUT_PARQUET)

writer = None

datasets = [
    ("ANAC", ANAC_CSV),
    ("BTS", BTS_CSV),
]

# barra de progresso GERAL
for nome, csv_path in tqdm(datasets, desc="Progresso geral (ANAC + BTS)", total=2):
    writer = csv_to_parquet_incremental(csv_path, writer)

# fechar parquet corretamente
writer.close()

print("\n✅ Unificação concluída com sucesso!")
print(f"📦 Arquivo gerado: {OUT_PARQUET}")
```

####Abrindo o parquet

O parquet ocupa pouco espaço mas na hora de abrir tem de descomprimir, reparei que so para abrir sem fazer o one hot encoding esta muito proximo de estourar a ram, o arquivo tem +35 milhoes de linhas qualaquer otimização e valida...

Os tipos de dados da colunas inteiras são do tipo int64 mas no nosso caso o int8 já supri nossa necessidade, ja que a coluna com maior range de valores e a 'scheduled_hour' hora do voo que vai da 0 a 23.

```python
import pandas as pd

df = pd.read_parquet("df_full_anac_bts_2021_01_2025_09.parquet")


# ================================
# DOWNCAST DE COLUNAS NUMÉRICAS
# ================================

cols_int8 = [
    "scheduled_hour",
    "hour_bucket",
    "is_peak_hour",
    "is_morning_peak",
    "is_evening_peak",
    "day_of_week",
    "is_weekend",
    "week_of_month",
    "month",
    "delayed",
]

for col in cols_int8:
    df[col] = df[col].astype("int8")

df.info(memory_usage="deep")


df.head()
```
